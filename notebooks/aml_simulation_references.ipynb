| Synthetic ground truth and benchmarking | SYNTHAML-2023; IBM-NEURIPS-SYNTHETIC-AML-2023; AMLSIM-GITHUB; SAML-D-KAGGLE | Positions this framework against existing synthetic AML benchmarks. |

````python
# Generate BibTeX starter entries from the reference table.
# Review and edit venue/authors before using these in a final manuscript.

def _sanitize_key(value):
    return ''.join(ch for ch in value if ch.isalnum()) or 'Reference'

def to_bibtex(ref):
    key = _sanitize_key(ref['key'])
    year = ref['year'] if str(ref['year']).isdigit() else 'n.d.'
    return (
        f"@misc{{{key},\n"
        f"  author = {{{ref['source']}}},\n"
        f"  title = {{{ref['title']}}},\n"
        f"  year = {{{year}}},\n"
        f"  url = {{{ref['url']}}},\n"
        f"  note = {{Accessed 2026-05-01}}\n"
        f"}}"
    )

print('\n\n'.join(to_bibtex(ref) for ref in references))
````
